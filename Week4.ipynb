You can in week one find more details explanations who the sample mean, sample standard deviation is calculated than here.

Sample Mean:

$$\bar{x} = \frac{1}{n} \sum^{n}_{i=1} x_i = \frac{1}{9} \cdot 24023 = \frac{24023}{9} = 3002.33 mm$$

Sample Standard Deviation (used programmel for this one. Again see week 1 for better example):

$$S^2 = \cfrac{1}{n-1} \sum^{n}_{i=1}(x_i - \bar{x}) = \cfrac{1}{8} \sum^{8}_{i=1}(x_i - 3002.33) = 13.75$$

$$S = \sqrt{S^2} = \sqrt{13.75} = 3.70$$

Standard Error of The Mean:

$$\sigma_{\bar{x}} = \frac{S}{\sqrt{n}} = \frac{3.70}{\sqrt{9}} = \frac{3.70}{3} = 1.23$$

The sample mean tells us that the average length of the samples is 3002.33, which is pretty good compared to the true mean of 3000. Remember that the mean is easily affected by outliers/extremes.

The Standard Deviation tells us how much the samples are deviated from the sample mean. With a std at 3.7mm, most concrete items will be within the range of approximately $3.7 * 2 = \pm 7.4mm$. The smaller std, the closer are the samples to the sample mean.

The SEM (Standard Error of the Mean) of 1.23 tells us that, if we repeatedly drew samples of size 9 from the population, the sample means would vary around the true mean with a standard deviation of about 1.23.


```python
items = np.array([3003, 3005, 2997, 3006, 2999, 2998, 3007, 3005, 3001])

print(f'Sample Mean: {np.mean(items)}')
print(f'Sample Variance: {np.var(items, ddof=1)}')
print(f'Sample Standard Deviation: {np.std(items, ddof=1)}')
```
